```python
def evidenceprobability():

    evidence = dict()
    for input_feature_name in data.columns[0:2]:
        evidence[input_feature_name] = dict()

        for diagnosis in [0.0, 1.0]:

            frequency = data[data[input_feature_name] == diagnosis].shape[0]
            p = (frequency / data.shape[0])
            evidence[input_feature_name][diagnosis] = p

    return evidence        
                                                                        
```
