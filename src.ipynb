```python
number_of_students = 200
coin_tosses = 5

coin_toss_results = np.random.binomial( n=coin_tosses, p=0.5, size=number_of_students)

print(coin_toss_results)
```

```python
rf = list()
for num_heads in range(0, coin_tosses + 1):
    frequency_num_heads = np.count_nonzero(coin_toss_results == num_heads)
    print(frequency_num_heads/number_of_students)

    print (rf)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
number_of_experiments = 40000
num_bernoulli_trials = 10

num_desired_outcomes = list()
relative_frequency = list()

experiment_result = np.random.binomial(n=num_bernoulli_trials, p=0.8, size=number_of_experiments)

print(experiment_result)
```

```python
experiment_result == 0
```

```python
for k in range(num_bernoulli_trials+1):
    frequency_num_heads = np.count_nonzero(experiment_result == k)
    rf = (frequency_num_heads/number_of_experiments)
    print(f"RF({k}) = {rf}")

    num_desired_outcomes.append(k)
    relative_frequency.append(rf)


    plt.xlabel("Number of people out of 10 people voting for BJP")
    plt.ylabel("Probabilty of k people out of 10 voting for BJP")
    plt.bar(num_desired_outcomes, relative_frequency)
```

```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
number_of_student = 400000000
mu = 170
sigma = 10

heights = np.random.normal(mu,sigma,size=number_of_student)
```

```python
print(heights)
```

```python
min_height = np.min(heights)
max_height = np.max(heights)

print(f"Min Height = {min_height}")
print(f"Max Height = {max_height}")
```

```python
num_bins = 1000
bin_width = (max_height - min_height)/num_bins
print(f"Bin Width = {bin_width}")
```

```python
plt.xlabel(f"{num_bins} bins from {min_height} to {max_height}")
plt.ylabel("Frequency of different bins")
plt.hist(heights, bins=num_bins)
```
